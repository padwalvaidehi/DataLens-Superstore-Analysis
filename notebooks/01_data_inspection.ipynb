```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

sales_by_category = df.groupby("Category")["Sales"].sum().sort_values()

ax = sales_by_category.plot(kind="barh", figsize=(9, 5))

ax.set_title("Total Sales by Product Category", fontsize=14)
ax.set_xlabel("Total Sales ($)")
ax.set_ylabel("Product Category")

# Format the x-axis in thousands
ax.xaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, _: f"${x / 1000:.0f}K")
)

# Display sales values at the end of each bar
for bar in ax.patches:
    ax.text(
        bar.get_width(),
        bar.get_y() + bar.get_height() / 2,
        f"${bar.get_width():,.0f}",
        va="center",
        ha="left",
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

```python
profit_by_category = df.groupby("Category")["Profit"].sum().sort_values()

ax = profit_by_category.plot(kind="barh", figsize=(9, 5))

ax.set_title("Total Profit by Product Category", fontsize=14)
ax.set_xlabel("Total Profit")
ax.set_ylabel("Product Category")

for bar in ax.patches:
    ax.text(
        bar.get_width(),
        bar.get_y() + bar.get_height() / 2,
        f"{bar.get_width():,.0f}",
        va="center",
        ha="left",
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

```python
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

category_analysis["Profit_Margin_%"] = (
    category_analysis["Total_Profit"]
    / category_analysis["Total_Sales"] * 100
).round(2)

category_analysis.sort_values(
    "Profit_Margin_%", ascending=False
)
```

```python
discount_analysis = df.groupby("Discount").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Order_Lines=("Order ID", "count")
)

discount_analysis["Profit_Margin_%"] = (
    discount_analysis["Total_Profit"]
    / discount_analysis["Total_Sales"] * 100
).round(2)

discount_analysis.sort_index()
```

```python
import matplotlib.pyplot as plt

discount_margin = discount_analysis["Profit_Margin_%"]

plt.figure(figsize=(9, 5))
plt.plot(
    discount_margin.index * 100,
    discount_margin.values,
    marker="o"
)

plt.axhline(y=0, linestyle="--")
plt.title("Profit Margin by Discount Level")
plt.xlabel("Discount (%)")
plt.ylabel("Profit Margin (%)")
plt.xticks(discount_margin.index * 100)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Sales and profit by product category

category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

category_analysis["Profit_Margin_%"] = (
    category_analysis["Total_Profit"] /
    category_analysis["Total_Sales"] * 100
).round(2)

category_analysis = category_analysis.sort_values(
    by="Total_Sales", ascending=False
)

category_analysis
```

```python
# Profit analysis by sub-category

subcategory_analysis = df.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

subcategory_analysis["Profit_Margin_%"] = (
    subcategory_analysis["Total_Profit"] /
    subcategory_analysis["Total_Sales"] * 100
).round(2)

subcategory_analysis = subcategory_analysis.sort_values(
    by="Total_Profit", ascending=True
)

subcategory_analysis
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.barh(
    subcategory_analysis["Sub-Category"],
    subcategory_analysis["Total_Profit"]
)

plt.axvline(x=0, linewidth=1)
plt.title("Profit by Product Sub-Category")
plt.xlabel("Total Profit")
plt.ylabel("Sub-Category")
plt.tight_layout()
plt.show()
```

```python
# Analyze profitability at different discount levels

discount_analysis = df.groupby("Discount").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Order_Lines=("Order ID", "count")
).reset_index()

discount_analysis["Profit_Margin_%"] = (
    discount_analysis["Total_Profit"] /
    discount_analysis["Total_Sales"] * 100
).round(2)

discount_analysis.sort_values("Discount")
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    discount_analysis["Discount"] * 100,
    discount_analysis["Profit_Margin_%"],
    marker="o"
)

plt.axhline(y=0, linewidth=1)
plt.title("Discount Level vs Profit Margin")
plt.xlabel("Discount (%)")
plt.ylabel("Profit Margin (%)")
plt.xticks(
    discount_analysis["Discount"] * 100,
    rotation=45
)
plt.tight_layout()
plt.show()
```

```python
# Monthly sales and profit analysis

df["Order Month"] = df["Order Date"].dt.to_period("M")

monthly_analysis = df.groupby("Order Month").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
).reset_index()

monthly_analysis
```

```python
# Identify the highest and lowest sales months

best_month = monthly_analysis.loc[
    monthly_analysis["Total_Sales"].idxmax()
]

worst_month = monthly_analysis.loc[
    monthly_analysis["Total_Sales"].idxmin()
]

print("Highest Sales Month:")
print(best_month)

print("\nLowest Sales Month:")
print(worst_month)
```

```python
import sqlite3

db_path = project_root / "data" / "processed" / "superstore.db"
conn = sqlite3.connect(db_path)

# Convert Period columns to strings for SQLite compatibility
df_sql = df.copy()

for col in df_sql.columns:
    if str(df_sql[col].dtype) == "period[M]":
        df_sql[col] = df_sql[col].astype(str)

# Load data into SQLite
df_sql.to_sql("superstore", conn, if_exists="replace", index=False)

print("SQL database created successfully!")
print("Rows loaded:", len(df_sql))
print("Table: superstore")
```

```python
# First SQL business query

query = """
SELECT
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    SUM(Quantity) AS Total_Quantity
FROM superstore;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    Category,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM superstore
GROUP BY Category
ORDER BY Total_Profit DESC;
"""

category_sql = pd.read_sql_query(query, conn)
category_sql
```

```python
query = """
SELECT
    "Sub-Category",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM superstore
GROUP BY "Sub-Category"
ORDER BY Total_Profit ASC
LIMIT 10;
"""

least_profitable = pd.read_sql_query(query, conn)
least_profitable
```

```python
query = """
SELECT
    Region,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM superstore
GROUP BY Region
ORDER BY Total_Profit DESC;
"""

region_sql = pd.read_sql_query(query, conn)
region_sql
```
